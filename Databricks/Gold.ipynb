
**Data Access**

```python
gold_loaction = 'abfss://gold@adeprojectstorageaccount.dfs.core.windows.net/'
silver_location = 'abfss://silver@adeprojectstorageaccount.dfs.core.windows.net/'
```

##Reading Data
**Trip Type Data**

```python
df_trip_type = spark.read.format('parquet')\
    .option('inferSchema', True)\
        .option('header',True)\
            .load(f'{silver_location}Trip_Type')
```

```python
df_trip_type.display()
```


**Trip Zone Data**

```python
df_trip_zone = spark.read.format('parquet')\
    .option('inferSchema', True)\
        .option('header',True)\
            .load(f'{silver_location}Trip_Zone')
```

```python
df_trip_zone.display()
```


**Trip Data**

```python
df_trip = spark.read.format('parquet')\
    .option('inferSchema', True)\
        .option('header',True)\
            .load(f'{silver_location}Trip_2025')
```

```python
df_trip.display()
```


#Loading


**Creating Databse**

```python
%sql
CREATE CATALOG gold;
```

```python
%sql

CREATE SCHEMA gold.data;
```

```python
df_trip_type.write.format('delta')\
    .mode('append')\
        .option('path',f'{gold_loaction}Trip_Type')\
            .saveAsTable('gold.data.trip_type')
```

```python
%sql

SELECT * FROM gold.data.trip_type
```

```python
df_trip_zone.write.format('delta')\
    .mode('append')\
        .option('path',f'{gold_loaction}Trip_Zone')\
            .saveAsTable('gold.data.trip_zone')
```

```python
%sql

SELECT
    *
FROM gold.data.trip_zone
```

```python
df_trip.write.format('delta')\
    .mode('append')\
        .option('path',f'{gold_loaction}Trip_2025')\
            .saveAsTable('gold.data.trip_data')
```

```python
%sql

UPDATE gold.data.trip_zone
SET trip_zone2 = 'unknown' 
WHERE trip_zone2 IS NULL
```

```python
%sql

SELECT * FROM gold.data.trip_zone
```

```python
%sql

DESCRIBE HISTORY gold.data.trip_zone
```

```python
%sql

RESTORE gold.data.trip_zone TO VERSION AS OF 0
```

```python
%sql

SELECT * FROM gold.data.trip_zone
```
